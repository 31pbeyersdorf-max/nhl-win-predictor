```python
games["goal_diff"] = games["goalsFor"] - games["goalsAgainst"]

print("Rows with a tie:", (games["goal_diff"] == 0).sum())
print(games["season"].value_counts().sort_index())
```

```python
games = games[games["goal_diff"] != 0].copy()
games["gameDate"] = pd.to_datetime(games["gameDate"].astype(str), format="%Y%m%d")
games["win"] = (games["goal_diff"] > 0).astype(int)

print(games.shape)
print(games["home_or_away"].value_counts())
```

```python
home = games[games["home_or_away"] == "HOME"]
print("Home win rate:", home["win"].mean())
print(home.groupby("season")["win"].mean().round(3))
```

```python
all_games = df[(df["situation"] == "all") & (df["playoffGame"] == 0)].copy()
all_games["gameDate"] = pd.to_datetime(all_games["gameDate"].astype(str), format="%Y%m%d")
all_games["goal_diff"] = all_games["goalsFor"] - all_games["goalsAgainst"]
all_games = all_games.sort_values(["team", "gameDate"]).reset_index(drop=True)
```

```python
stats = ["xGoalsFor", "xGoalsAgainst", "goalsFor", "goalsAgainst",
         "shotsOnGoalFor", "shotsOnGoalAgainst",
         "highDangerShotsFor", "highDangerShotsAgainst"]

for col in stats:
    all_games[col + "_avg10"] = (
        all_games.groupby("team")[col]
        .transform(lambda s: s.shift(1).rolling(10, min_periods=5).mean())
    )

all_games["rest_days"] = all_games.groupby("team")["gameDate"].diff().dt.days.clip(upper=7)
```

```python
feat_cols = [c for c in all_games.columns if c.endswith("_avg10")] + ["rest_days"]

home = all_games[all_games["home_or_away"] == "HOME"][["gameId", "season", "gameDate", "team", "goal_diff"] + feat_cols]
away = all_games[all_games["home_or_away"] == "AWAY"][["gameId", "team"] + feat_cols]

home = home.rename(columns={**{c: "home_" + c for c in feat_cols}, "team": "home_team"})
away = away.rename(columns={**{c: "away_" + c for c in feat_cols}, "team": "away_team"})

model_df = home.merge(away, on="gameId")
model_df = model_df[model_df["goal_diff"] != 0].copy()   # drop shootouts
model_df["home_win"] = (model_df["goal_diff"] > 0).astype(int)
model_df = model_df.dropna()

print(model_df.shape)
model_df[["gameDate", "home_team", "away_team", "home_xGoalsFor_avg10", "away_xGoalsFor_avg10", "home_win"]].head()
```

```python
print(all_games[all_games["team"] == "NYR"][["gameDate", "xGoalsFor", "xGoalsFor_avg10"]].head(8))
```

```python
model_df = model_df[model_df["season"] <= 2025]   # drop the partial 2026 season

train = model_df[model_df["season"] <= 2021]
test = model_df[model_df["season"] >= 2022]

X_cols = ["home_" + c for c in feat_cols] + ["away_" + c for c in feat_cols]

X_train, y_train = train[X_cols], train["home_win"]
X_test, y_test = test[X_cols], test["home_win"]

print("Train games:", len(train), "| Test games:", len(test))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, log_loss

model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)

probs = model.predict_proba(X_test)[:, 1]
preds = (probs > 0.5).astype(int)
```

```python
base_rate = y_train.mean()

print("Baseline (always home) accuracy:", round(y_test.mean(), 4))
print("Model accuracy:                 ", round(accuracy_score(y_test, preds), 4))
print()
print("Baseline log loss:", round(log_loss(y_test, [base_rate] * len(y_test)), 4))
print("Model log loss:   ", round(log_loss(y_test, probs), 4))
```
